## Part 1: Action Designator (E1–E6)

### E1 · The Gap & Whole-Course Judgment

Reaching the intended final state is not sufficient to demonstrate successful task completion. Although the robot executes a smooth trajectory and places the mug on the table, it spills milk across the counter and leaves the fridge open.

The robot therefore achieves the nominal end state but fails at whole-course judgment. A successful solution to the Body Motion Problem must satisfy the intended goal while ensuring that the entire sequence of actions does not leave the environment in an unacceptable state.

The two violated whole-course constraints are:

1. **Spill prevention:** The robot must prevent milk from spilling during retrieval and pouring.
2. **Environment restoration:** The robot must leave the fridge closed after retrieving the carton.

### E2 · Constrained Action Designator & Grounding

**Working Interpretation:**

`(:type serving) (:recipient Eva) (:substance whole-milk)`

**Inferred Function/Means:**

`(:vessel-role drinking-mug)`

**Grounded Physical Object:**

`(:object Carton-A)`

The working interpretation represents Eva's intended task of receiving a drink. The robot infers that the mug is the appropriate vessel for serving the drink. It then grounds the intended action to the actual physical object, Carton A, which contains opened whole milk.

The selection of whole milk is based on the specified episode involving Carton A and its wet surface. Since Eva has not explicitly stated a preference between whole milk and oat drink, the robot should confirm her preference before committing to the selection.

### E3 · The Commitment Decision & Asymmetric Failure Costs

The robot should choose the 5 mm test lift near the shelf before committing to the high carry because Carton A has a wet surface that creates uncertainty about grasp stability.

**Predicted effect:** The test lift allows the robot to observe whether the carton remains securely held by the gripper.

**Cost of failure:** Failure during the low test lift has a relatively small cost because the carton remains close to the shelf. However, failure during a high carry could cause the carton to fall, spill milk and contaminate the kitchen environment.

**Evidence:** The observed downward movement of the carton relative to the gripper during the 5 mm test lift indicates that the initial grasp is slipping.

Therefore, the robot should detect the slippage, adapt its grasp and verify the new grasp before committing to the high carry.

### E4 · Causal Prediction across Material Variants

The stopping rule must depend on the material being poured and its flow behaviour.

**1. Milk into coffee**

`(:stop-when (coffee-volume-reaches-target))`

Milk has relatively low viscosity and flows easily. Therefore, the robot should begin tilting its wrist back slightly before the desired final volume is reached. This compensates for the delay between initiating the stopping movement and the actual cessation of liquid flow.

**2. Pancake batter into a pan**

`(:stop-when (target-batter-portion-reached))`

Pancake batter has higher viscosity and flows more slowly than milk. The robot must account for its slower flow and different deformation behaviour. It should regulate the wrist tilt based on the observed batter flow and begin tilting back when the target portion is approached.

Material dynamics influence when the robot must command the wrist to tilt back. A single fixed stopping time or wrist angle is insufficient for different materials. The robot must predict and monitor the actual flow behaviour.

### E5 · System Decomposition & The Reasoning Loop

The carton slip event can be traced through the seven-stage reasoning loop:

1. **Infer:** The robot interprets Eva's request as serving her a drink and develops the corresponding serving intention.

2. **Ground:** The robot identifies the physical milk carton and the intended drinking mug.

3. **Predict:** The robot predicts the possible consequences of lifting the carton, including the risk of slipping due to its wet surface.

4. **Decide:** The robot chooses a low 5 mm test lift instead of immediately performing a high carry because the failure cost is lower.

5. **Regulate:** The robot executes and controls the small lifting movement.

6. **Detect & Revise:** When downward movement of the carton relative to the gripper is observed, the robot detects grasp slippage. This triggers the Detect & Revise stage. The original action designator is revised to include an adapted grasp configuration and another test lift.

7. **Verify:** The robot performs another test to establish that the adapted grasp is secure before committing to the high carry and subsequent pouring action.

The important point is that the robot does not blindly continue executing its original intention. It uses observations to revise its action designator and adapt its behaviour.

### E6 · Conceptual Ablation & Evidence of Competence

Removing the "Check before commitment" capability makes the robot less adaptable and more vulnerable to environmental variations.

Consider a kitchen environment where Carton A has a wet, slippery surface. The ablated robot applies its fixed default high-side-pinch grasp and immediately performs a high lift without testing grasp stability. Due to reduced friction, the carton may slip from the gripper, causing it to fall and spill milk.

The original system would perform a low test lift, detect the slipping behaviour, revise its grasp and verify the adapted grasp before committing to the high carry.

A successful execution in a single environment does not establish general competence. It only demonstrates that the robot succeeded under one particular set of conditions. Evidence of general competence requires successful performance across different task and environmental variants, including changes in object properties, surface conditions and grasping constraints.

Therefore, removing the checking capability reduces the robot's ability to detect failures, adapt its actions and demonstrate reliable competence across different environments.

## E1 · Opening Judgment (Data vs. Competence)

No. A robot successfully serving coffee 100 times in a row does not prove that it can perform the task in any home environment.

The 100 successful demonstrations provide evidence that the robot can perform the task under the specific conditions in which it was tested. However, different homes may have different layouts, obstacles, lighting conditions, furniture arrangements, mug shapes, and floor surfaces. Changes in these conditions may affect perception, grasping, navigation, and execution.

A demonstration video also shows the outcome but does not necessarily reveal whether the robot independently understood the situation, adapted to unexpected changes, or checked the safety of its actions.

Therefore, successful demonstrations are evidence of performance in tested conditions, not proof of general competence.

**Confidence: 5/5**

## E2 · Changed Case Prediction (Physical Bounds)

If the robot successfully carries a half-full white mug but then attempts to carry a nearly full mug along the same route, the liquid is more likely to spill during a turn.

When the robot turns, the liquid moves due to inertia. A higher fill level leaves less empty space inside the mug, increasing the risk of liquid reaching the rim and spilling. The robot may also need to adjust its acceleration, turning speed, and movement trajectory.

The untested assumption in the original run is that successful transportation of a half-full mug guarantees successful transportation at a higher fill level.

The robot should not assume that the same movement parameters are safe for both conditions. It should account for the fill level and predict the consequences before execution.

## E3 · The Commitment Decision (Warrant Gate)

The robot should **ask Eva** whether using the white mug is acceptable instead of the usual blue mug.

The blue mug is unavailable, and there is no confirmed evidence that Eva accepts the white mug as a substitute. Although the robot can physically use the white mug, physical feasibility alone does not establish that the proposed action satisfies Eva's preference.

**Evidence required:** Explicit confirmation from Eva that she accepts the white mug.

After receiving confirmation, the robot can update its goal and proceed with the remaining feasibility and safety checks.

## E4 · The Five Questions of Understanding (T1–T5)

1. **Explain (T1):** Why did the robot choose the white mug instead of the blue mug, and what information did it use to make that decision?

2. **Predict (T2):** What is likely to happen to the coffee when the robot carries a nearly full mug through a sharp turn?

3. **Design (T3):** How should the robot plan its grasping, carrying speed, and navigation trajectory to deliver the coffee without spilling it?

4. **Bound (T4):** What is the maximum safe turning speed or acceleration for carrying a nearly full mug without exceeding the acceptable liquid movement or spilling limits?

5. **Check (T5):** What sensor observations and execution records can confirm that the coffee was delivered successfully without spilling or violating safety constraints?

The Predict and Bound questions address changed physical conditions that a demonstration video alone cannot establish. They require a model of the physical system and evidence from relevant tests.

## E5 · System Decomposition (Mapping to Architecture)

The robot requires the following functional modules to handle Eva's request:

| Functional module | Responsibility |
|---|---|
| Interpretation System | Understands Eva's natural language request and converts it into a concrete goal, including the target object and destination. |
| Perception System | Processes sensor observations to identify the mug, its fill level, Eva's position, and environmental conditions. |
| Proposer | Generates possible actions and execution parameters, such as grasp position, carrying speed, and tilt rate. |
| Semantic Digital Twin | Predicts the physical consequences of proposed actions, including forces, liquid movement, and possible failures. |
| Warrant Gate | Checks whether the proposed action is grounded, feasible, safe, monitorable, repairable, and recordable before allowing execution. |
| Execution System | Sends approved commands to the robot's actuators to grasp and transport the mug. |
| Recording / Audit System | Records the request, internal belief, observations, approved or rejected proposal, and execution outcome for later analysis. |

**Data flow in the Integration Diagram:**

Eva's natural language request first enters the Interpretation System, which converts it into a structured goal. The Perception System gathers information from the robot's sensors and updates its internal belief about the environment.

The Proposer generates a candidate action based on the goal and current belief. The Semantic Digital Twin predicts the expected physical consequences of that action. The Warrant Gate evaluates the proposal against the required criteria.

If the proposal fails any important check, it is rejected and another proposal can be generated. If all checks pass, the Execution System sends commands to the robot's actuators. Finally, the Recording System stores the observations, decisions, and actual execution outcome.

This creates a closed-loop architecture in which the robot observes, reasons, predicts, checks, acts, and records rather than blindly executing a fixed sequence of actions.

## E6 · Transfer & Exit Ticket (Justified Adaptation)

If Eva stands up while the robot is carrying the coffee, the robot should pause its current navigation trajectory and update its estimate of Eva's position. It should then plan a revised route towards her new position, taking into account her movement and the surrounding obstacles.

A possible failure mechanism is that the robot continues towards Eva's previous position or makes a sudden movement while changing direction. This could cause a collision, destabilize the mug, or spill the coffee due to excessive acceleration or liquid movement.

**Evidence criterion before committing:** The robot must obtain an updated sensor observation confirming Eva's current position and verify that the revised trajectory is collision-free and that its predicted forces and liquid movement remain within the specified safety limits.

The robot should only commit to the revised path after the Warrant Gate confirms that the action satisfies the required checks. It should continue monitoring Eva's position during execution and replan if the situation changes again.

```python
#!/usr/bin/env python3
"""
Integrated Intelligent Systems – Tutorial 1
Embodied Cognitive Architecture

This is the consolidated Python implementation for the tutorial tasks.
Run this whole file / paste it into one Jupyter Code cell.
"""

from dataclasses import dataclass
from typing import Dict, Any, Tuple, Optional, List
import random
import time
import statistics


# ============================================================
# 1. DATA STRUCTURES
# ============================================================

@dataclass(frozen=True)
class GoalSpecification:
    """Concrete goal derived from the user's instruction and context."""
    target_object: str
    destination: str
    target_volume_ml: float
    max_allowable_force_n: float
    max_slosh_wave_mm: float
    max_nav_velocity_mps: float
    time_frame_seconds: float


@dataclass
class ParameterPacket:
    """Candidate robot motion parameters generated by the proposer."""
    grasp_pose: Tuple[float, float, float]
    max_velocity_mps: float
    tilt_rate_radps: float
    # Task 2.1: contact depth is used to model bottoming-out.
    contact_depth_m: float = 0.01
    source_generator: str = "stochastic_sampler"


@dataclass
class Observation:
    """Noisy measurements received from the simulated sensors."""
    timestamp: float
    measured_temperature_c: float
    measured_friction: float
    eva_position: str
    room_busyness: str


@dataclass
class BeliefState:
    """The robot's current internal estimate of the world."""
    pouch_present: bool = True
    pouch_material: str = "flexible_plastic"
    fluid_viscosity_pas: float = 1.2
    vial_ready: bool = True
    camera_active: bool = True
    fallback_path_exists: bool = True
    audit_logger_enabled: bool = True

    ambient_temperature_c: float = 21.5
    surface_friction_coefficient: float = 0.45
    eva_position: str = "living_room"
    room_busyness: str = "medium"
    preferred_cup: str = "blue_ceramic_mug"

    # Task 4.1: estimated remaining battery charge.
    battery_level_pct: float = 100.0


# ============================================================
# 2. SYSTEM 1 – PROPOSER
# ============================================================

class Proposer:
    """
    Generates candidate actions using Gaussian sampling.
    Gaussian sampling explores around a reasonable mean while
    producing mostly similar, but not identical, proposals.
    """

    def __init__(
        self,
        velocity_mean: float = 0.45,
        velocity_std: float = 0.08,
        tilt_mean: float = 0.08,
        tilt_std: float = 0.02
    ):
        self.velocity_mean = velocity_mean
        self.velocity_std = velocity_std
        self.tilt_mean = tilt_mean
        self.tilt_std = tilt_std

    def propose(self) -> ParameterPacket:
        velocity = max(
            0.05,
            random.gauss(self.velocity_mean, self.velocity_std)
        )
        tilt = max(
            0.01,
            random.gauss(self.tilt_mean, self.tilt_std)
        )

        return ParameterPacket(
            grasp_pose=(
                random.uniform(0.40, 0.50),
                random.uniform(0.10, 0.15),
                random.uniform(0.25, 0.35)
            ),
            max_velocity_mps=round(velocity, 3),
            tilt_rate_radps=round(tilt, 3),
            contact_depth_m=0.01,
            source_generator="gaussian_sampler"
        )


# ============================================================
# 3. INTERPRETATION, SENSING AND PERCEPTION
# ============================================================

class InterpretationSystem:
    """Maps a natural-language instruction into a formal goal."""

    def __init__(self, knowledge_base: Dict[str, Any]):
        self.kb = knowledge_base

    def interpret(
        self,
        raw_utterance: str,
        belief: BeliefState
    ) -> GoalSpecification:

        print(
            f"\n[INTERPRETATION] Analyzing: '{raw_utterance}' "
            f"(Eva: {belief.eva_position}, "
            f"room busyness: {belief.room_busyness})"
        )

        cup_type = belief.preferred_cup
        cup_properties = self.kb["cups"].get(
            cup_type,
            {"max_force": 5.0, "max_wave": 2.0}
        )

        busyness_factor = self.kb["room_busyness_limits"].get(
            belief.room_busyness, 0.5
        )
        max_nav_velocity = 1.2 * busyness_factor

        is_urgent = (
            "schnell" in raw_utterance.lower()
            or "fast" in raw_utterance.lower()
        )
        time_frame = 30.0 if is_urgent else 60.0
        target_volume = 12.0 if is_urgent else 15.0

        goal = GoalSpecification(
            target_object=cup_type,
            destination=belief.eva_position,
            target_volume_ml=target_volume,
            max_allowable_force_n=cup_properties["max_force"],
            max_slosh_wave_mm=cup_properties["max_wave"],
            max_nav_velocity_mps=max_nav_velocity,
            time_frame_seconds=time_frame
        )

        print(
            f" -> Goal: object={goal.target_object}, "
            f"max force={goal.max_allowable_force_n} N, "
            f"max wave={goal.max_slosh_wave_mm} mm, "
            f"max navigation speed={goal.max_nav_velocity_mps} m/s"
        )
        return goal


class SensingSystem:
    """Simulates noisy physical sensors."""

    def sense(
        self,
        true_ambient_temp: float,
        true_friction: float,
        eva_pos: str,
        busyness: str
    ) -> Observation:

        noisy_temp = true_ambient_temp + random.uniform(-0.5, 0.5)
        noisy_friction = true_friction + random.uniform(-0.05, 0.05)

        return Observation(
            timestamp=time.time(),
            measured_temperature_c=round(noisy_temp, 2),
            measured_friction=round(noisy_friction, 2),
            eva_position=eva_pos,
            room_busyness=busyness
        )


class PerceptionSystem:
    """Updates the robot's internal belief from sensor observations."""

    def __init__(self, belief_state: BeliefState):
        self.belief_state = belief_state

    def update(self, observation: Observation) -> None:
        self.belief_state.ambient_temperature_c = (
            observation.measured_temperature_c
        )
        self.belief_state.surface_friction_coefficient = (
            observation.measured_friction
        )
        self.belief_state.eva_position = observation.eva_position
        self.belief_state.room_busyness = observation.room_busyness


# ============================================================
# 4. SYSTEM 2 – SEMANTIC DIGITAL TWIN
# ============================================================

class SemanticDigitalTwin:
    """
    Predicts physical outcomes using an imperfect model.
    Random prediction errors represent the sim-to-real gap.
    """

    def __init__(self, belief_state: BeliefState):
        self.belief = belief_state

    def predict_properties(
        self,
        params: ParameterPacket,
        goal: GoalSpecification
    ) -> Dict[str, float]:

        # Task 2.1: base stiffness depends on surface friction.
        stiffness_coefficient = 12.5 * (
            1.0 + self.belief.surface_friction_coefficient * 0.1
        )

        # Task 2.1: after 0.02 m contact depth, stiffness rises
        # exponentially by a factor of 1.8 for each additional 0.02 m.
        if params.contact_depth_m > 0.02:
            excess_depth = params.contact_depth_m - 0.02
            stiffness_coefficient *= 1.8 ** (excess_depth / 0.02)

        predicted_force = stiffness_coefficient * (
            params.max_velocity_mps ** 1.5
        )

        # Task 3.1: viscosity decreases by 1% per degree above 20°C.
        temperature_excess = max(
            0.0,
            self.belief.ambient_temperature_c - 20.0
        )
        viscosity_factor = 1.0 - temperature_excess * 0.01
        effective_viscosity = (
            self.belief.fluid_viscosity_pas * viscosity_factor
        )
        effective_viscosity = max(0.01, effective_viscosity)

        radius = 0.05
        wave_height_m = (
            params.tilt_rate_radps * radius
        ) / effective_viscosity
        predicted_wave = wave_height_m * 1000.0

        # Epistemic uncertainty: the digital twin is not perfectly accurate.
        epistemic_force_error = random.uniform(-0.2, 0.3)
        epistemic_wave_error = random.uniform(-0.1, 0.2)

        return {
            "predicted_force_n": round(
                max(0.0, predicted_force + epistemic_force_error), 2
            ),
            "predicted_wave_mm": round(
                max(0.0, predicted_wave + epistemic_wave_error), 2
            ),
            "predicted_nav_velocity": min(
                params.max_velocity_mps,
                goal.max_nav_velocity_mps
            ),
            "effective_viscosity_pas": round(effective_viscosity, 3)
        }


# ============================================================
# 5. WARRANT GATE – SAFETY AND FEASIBILITY CHECKS
# ============================================================

class WarrantGate:
    """Checks whether a proposed action is justified before execution."""

    def __init__(self, sdt: SemanticDigitalTwin):
        self.sdt = sdt

    def validate(
        self,
        params: ParameterPacket,
        goal: GoalSpecification,
        belief: BeliefState
    ) -> Dict[str, Any]:

        predictions = self.sdt.predict_properties(params, goal)

        # Task 4.1: simplified estimated battery drain model.
        # This is an implementation assumption because the task gives
        # the 15% threshold but does not provide a drain equation.
        estimated_battery_drain = (
            10.0
            + 10.0 * params.max_velocity_mps
            + 5.0 * params.tilt_rate_radps
        )
        predictions["estimated_battery_drain_pct"] = round(
            estimated_battery_drain, 2
        )

        checks = {
            "grounded": belief.vial_ready and belief.pouch_present,
            "feasible": (
                params.max_velocity_mps > 0.0
                and params.tilt_rate_radps > 0.0
            ),
            "safe": (
                predictions["predicted_force_n"]
                <= goal.max_allowable_force_n
                and predictions["predicted_wave_mm"]
                <= goal.max_slosh_wave_mm
                and predictions["predicted_nav_velocity"]
                <= goal.max_nav_velocity_mps
            ),
            "monitorable": belief.camera_active,
            "repairable": belief.fallback_path_exists,
            "recordable": belief.audit_logger_enabled,
            "energetically_viable": (
                estimated_battery_drain < 15.0
                and belief.battery_level_pct >= estimated_battery_drain
            )
        }

        all_passed = all(checks.values())
        failed_criteria = [
            name for name, passed in checks.items() if not passed
        ]

        return {
            "passed": all_passed,
            "failed_criterion": (
                failed_criteria[0] if failed_criteria else None
            ),
            "failed_criteria": failed_criteria,
            "predictions": predictions,
            "warrant_checks": checks
        }


# ============================================================
# 6. EXECUTION AND EPISTEMIC AUDIT TRAIL
# ============================================================

class ExecutionEngine:
    """Executes approved actions and records the evidence trail."""

    def __init__(self):
        self.audit_trail: List[Dict[str, Any]] = []

    def execute_action(
        self,
        params: ParameterPacket,
        goal: GoalSpecification,
        warrant_result: Dict[str, Any]
    ) -> Dict[str, Any]:

        print(
            f"\n[ACTUATOR] Executing: "
            f"v_max={params.max_velocity_mps} m/s"
        )
        time.sleep(0.1)

        predictions = warrant_result["predictions"]
        actual_force = (
            predictions["predicted_force_n"]
            + random.uniform(-0.3, 0.6)
        )
        actual_wave = (
            predictions["predicted_wave_mm"]
            + random.uniform(-0.2, 0.5)
        )

        spill_occurred = (
            actual_force > goal.max_allowable_force_n
            or actual_wave > goal.max_slosh_wave_mm
        )
        status = "FAILURE" if spill_occurred else "SUCCESS"

        return {
            "actual_force_n": round(actual_force, 2),
            "actual_wave_mm": round(actual_wave, 2),
            "actual_volume_transferred_ml": (
                goal.target_volume_ml
                if not spill_occurred
                else goal.target_volume_ml - 3.0
            ),
            "spill_occurred": spill_occurred,
            "status": status
        }

    def record_trace(
        self,
        asked_utterance: str,
        goal: GoalSpecification,
        belief: BeliefState,
        proposal: ParameterPacket,
        warrant_result: Dict[str, Any],
        observation: Observation,
        committed: bool,
        execution_outcome: Optional[Dict[str, Any]] = None
    ) -> None:

        record = {
            "timestamp": time.time(),
            "asked": {
                "raw_utterance": asked_utterance,
                "interpreter_output": {
                    "target": goal.target_object,
                    "max_force_limit_n": goal.max_allowable_force_n,
                    "max_slosh_limit_mm": goal.max_slosh_wave_mm,
                    "max_nav_velocity": goal.max_nav_velocity_mps
                }
            },
            "belief": {
                "eva_position": belief.eva_position,
                "room_busyness": belief.room_busyness,
                "ambient_temp_c": belief.ambient_temperature_c,
                "surface_friction": belief.surface_friction_coefficient,
                "battery_level_pct": belief.battery_level_pct
            },
            "observed": {
                "timestamp": observation.timestamp,
                "measured_temp_c": observation.measured_temperature_c,
                "measured_friction": observation.measured_friction,
                "eva_position": observation.eva_position,
                "room_busyness": observation.room_busyness
            },
            "committed": {
                "action_approved": committed,
                "proposed_parameters": {
                    "max_velocity_mps": proposal.max_velocity_mps,
                    "tilt_rate_radps": proposal.tilt_rate_radps,
                    "contact_depth_m": proposal.contact_depth_m
                },
                "warrant_verdict": warrant_result
            },
            "execution_outcome": execution_outcome
        }

        self.audit_trail.append(record)


# ============================================================
# 7. TASK 5.1 – AUDIT ANALYZER
# ============================================================

def audit_analyzer(audit_trail: List[Dict[str, Any]]) -> Dict[str, Any]:
    """
    Summarizes rejected/approved proposals, common failed checks,
    and the difference between predicted and observed execution force.
    """

    approved_count = 0
    rejected_count = 0
    failed_criterion_counts: Dict[str, int] = {}
    absolute_force_errors = []
    squared_force_errors = []

    for trace in audit_trail:
        committed = trace["committed"]
        warrant = committed["warrant_verdict"]
        approved = committed["action_approved"]

        if approved:
            approved_count += 1
        else:
            rejected_count += 1

        for criterion in warrant.get("failed_criteria", []):
            failed_criterion_counts[criterion] = (
                failed_criterion_counts.get(criterion, 0) + 1
            )

        outcome = trace.get("execution_outcome")
        if outcome is not None:
            predicted_force = warrant["predictions"]["predicted_force_n"]
            actual_force = outcome["actual_force_n"]
            error = actual_force - predicted_force
            absolute_force_errors.append(abs(error))
            squared_force_errors.append(error ** 2)

    most_frequent_failed_criterion = (
        max(failed_criterion_counts, key=failed_criterion_counts.get)
        if failed_criterion_counts else None
    )

    return {
        "approved_count": approved_count,
        "rejected_count": rejected_count,
        "failed_criterion_counts": failed_criterion_counts,
        "most_frequent_failed_criterion": most_frequent_failed_criterion,
        "mean_absolute_force_error_n": (
            round(statistics.mean(absolute_force_errors), 3)
            if absolute_force_errors else None
        ),
        "mean_squared_force_error_n2": (
            round(statistics.mean(squared_force_errors), 3)
            if squared_force_errors else None
        ),
        "force_error_variance": (
            round(statistics.pvariance(absolute_force_errors), 3)
            if len(absolute_force_errors) > 1 else None
        )
    }


# ============================================================
# 8. MAIN DEMONSTRATION
# ============================================================

if __name__ == "__main__":

    random.seed(7)  # Makes this demonstration reproducible.

    knowledge_base = {
        "cups": {
            "blue_ceramic_mug": {
                "max_force": 4.5,
                "max_wave": 1.5
            },
            "plastic_cup": {
                "max_force": 7.0,
                "max_wave": 3.0
            }
        },
        "room_busyness_limits": {
            "low": 1.0,
            "medium": 0.6,
            "high": 0.3
        }
    }

    # Initialize all architecture components.
    belief_state = BeliefState(
        room_busyness="medium",
        preferred_cup="blue_ceramic_mug"
    )
    sensing = SensingSystem()
    perception = PerceptionSystem(belief_state)
    interpreter = InterpretationSystem(knowledge_base)
    proposer = Proposer()
    sdt = SemanticDigitalTwin(belief_state)
    warrant_gate = WarrantGate(sdt)
    executive = ExecutionEngine()

    # Step 1: sensing and perception.
    print("=== STEP 1: SENSING AND PERCEPTION ===")
    raw_observation = sensing.sense(
        true_ambient_temp=21.5,
        true_friction=0.45,
        eva_pos="living_room",
        busyness="medium"
    )
    perception.update(raw_observation)
    print("Raw sensor observation:", raw_observation)
    print(
        "Updated belief:",
        f"temperature={belief_state.ambient_temperature_c}°C,",
        f"friction={belief_state.surface_friction_coefficient}"
    )

    # Step 2: interpretation.
    print("\n=== STEP 2: INSTRUCTION INTERPRETATION ===")
    raw_command = "Bring me my coffee fast!"
    goal = interpreter.interpret(raw_command, belief_state)

    # Step 3: Gaussian proposer generates a candidate.
    print("\n=== STEP 3: GAUSSIAN PROPOSER ===")
    sampled_proposal = proposer.propose()
    print("Sampled proposal:", sampled_proposal)

    # Step 4: deliberately aggressive proposal – expected rejection.
    print("\n=== TASK 2.2: AGGRESSIVE PROPOSAL TEST ===")
    aggressive_proposal = ParameterPacket(
        grasp_pose=(0.45, 0.12, 0.30),
        max_velocity_mps=0.35,
        tilt_rate_radps=0.08,
        contact_depth_m=0.05,
        source_generator="aggressive_test"
    )
    aggressive_result = warrant_gate.validate(
        aggressive_proposal, goal, belief_state
    )
    print("Predictions:", aggressive_result["predictions"])
    print("Checks:", aggressive_result["warrant_checks"])
    print(
        "Verdict:",
        "APPROVED" if aggressive_result["passed"]
        else f"REJECTED – {aggressive_result['failed_criteria']}"
    )
    executive.record_trace(
        raw_command, goal, belief_state, aggressive_proposal,
        aggressive_result, raw_observation,
        committed=False
    )

    # Step 5: temperature and slosh test from Task 3.2.
    print("\n=== TASK 3.2: HIGH-TEMPERATURE / HIGH-TILT TEST ===")
    original_temperature = belief_state.ambient_temperature_c
    belief_state.ambient_temperature_c = 28.0

    hot_proposal = ParameterPacket(
        grasp_pose=(0.45, 0.12, 0.30),
        max_velocity_mps=0.20,
        tilt_rate_radps=0.15,
        contact_depth_m=0.01,
        source_generator="thermal_test"
    )
    hot_result = warrant_gate.validate(
        hot_proposal, goal, belief_state
    )
    print("Predictions:", hot_result["predictions"])
    print("Checks:", hot_result["warrant_checks"])
    print(
        "Verdict:",
        "APPROVED" if hot_result["passed"]
        else f"REJECTED – {hot_result['failed_criteria']}"
    )
    executive.record_trace(
        raw_command, goal, belief_state, hot_proposal,
        hot_result, raw_observation,
        committed=False
    )

    # Restore the normal temperature before the next test.
    belief_state.ambient_temperature_c = original_temperature

    # Step 6: battery test from Task 4.2.
    print("\n=== TASK 4.2: LOW-BATTERY TEST ===")
    original_battery = belief_state.battery_level_pct
    belief_state.battery_level_pct = 5.0

    low_battery_proposal = ParameterPacket(
        grasp_pose=(0.45, 0.12, 0.30),
        max_velocity_mps=0.20,
        tilt_rate_radps=0.02,
        contact_depth_m=0.01,
        source_generator="low_battery_test"
    )
    battery_result = warrant_gate.validate(
        low_battery_proposal, goal, belief_state
    )
    print("Predictions:", battery_result["predictions"])
    print("Checks:", battery_result["warrant_checks"])

    if not battery_result["warrant_checks"]["energetically_viable"]:
        print("Decision: REQUEST_RECHARGE")
    else:
        print("Decision: COFFEE_DELIVERY")

    executive.record_trace(
        raw_command, goal, belief_state, low_battery_proposal,
        battery_result, raw_observation,
        committed=False
    )
    belief_state.battery_level_pct = original_battery

    # Step 7: safe candidate and conditional execution.
    print("\n=== STEP 7: SAFE CANDIDATE AND WARRANT GATE ===")
    safe_proposal = ParameterPacket(
        grasp_pose=(0.45, 0.12, 0.30),
        max_velocity_mps=0.45,
        tilt_rate_radps=0.02,
        contact_depth_m=0.01,
        source_generator="safe_resampled_proposal"
    )
    safe_result = warrant_gate.validate(
        safe_proposal, goal, belief_state
    )
    print("Predictions:", safe_result["predictions"])
    print("Checks:", safe_result["warrant_checks"])

    if safe_result["passed"]:
        print("Gate verdict: APPROVED")
        final_outcome = executive.execute_action(
            safe_proposal, goal, safe_result
        )
        executive.record_trace(
            raw_command, goal, belief_state, safe_proposal,
            safe_result, raw_observation,
            committed=True,
            execution_outcome=final_outcome
        )
        print("Execution outcome:", final_outcome)
    else:
        print(
            "Gate verdict: REJECTED –",
            safe_result["failed_criteria"]
        )
        executive.record_trace(
            raw_command, goal, belief_state, safe_proposal,
            safe_result, raw_observation,
            committed=False
        )

    # Step 8: print audit trail and Task 5.1 analysis.
    print(
        f"\n=== COMPLETE AUDIT TRAIL "
        f"({len(executive.audit_trail)} ENTRIES) ==="
    )
    for index, trace in enumerate(executive.audit_trail, start=1):
        approved = trace["committed"]["action_approved"]
        outcome = trace.get("execution_outcome")
        outcome_text = (
            f" | Outcome: {outcome['status']} "
            f"(spill={outcome['spill_occurred']})"
            if outcome else ""
        )
        print(
            f"Entry {index}: approved={approved}"
            f"{outcome_text} | "
            f"command='{trace['asked']['raw_utterance']}'"
        )

    print("\n=== TASK 5.1: AUDIT ANALYSIS ===")
    analysis = audit_analyzer(executive.audit_trail)
    for key, value in analysis.items():
        print(f"{key}: {value}")


# ============================================================
# WRITTEN ANSWER NOTES FOR THE TUTORIAL
# ============================================================
#
# Task 1.1 – Architecture mapping:
# - InterpretationSystem.interpret: converts the utterance/context into GoalSpecification.
# - SensingSystem.sense: obtains noisy observations from the environment.
# - PerceptionSystem.update: updates the internal BeliefState.
# - Proposer.propose: generates candidate ParameterPacket actions.
# - SemanticDigitalTwin.predict_properties: predicts force, slosh and navigation speed.
# - WarrantGate.validate: checks grounding, feasibility, safety, monitoring,
#   repairability, recordability and energy viability.
# - ExecutionEngine.execute_action: executes an approved action in the simulation.
# - ExecutionEngine.record_trace: records the asked, belief, observed,
#   committed and execution-outcome information.
#
# Task 1.2 – Why Gaussian sampling?
# Gaussian sampling explores around a chosen mean while keeping most samples
# near that mean. Its standard deviation controls exploration. It is a useful
# proposer, but it does not itself guarantee safety; the Warrant Gate checks
# each candidate before execution.
#
# Task 2.1 – Contact depth:
# The implementation uses 0.02 m as the bottoming-out threshold and multiplies
# stiffness by 1.8 for every additional 0.02 m of contact depth.
#
# Task 2.2 – Aggressive proposal:
# The test uses 0.35 m/s and 0.05 m contact depth. The increased stiffness
# raises predicted force and should cause the safety check to reject it.
#
# Task 3.1 – Temperature:
# Above 20°C, effective viscosity decreases by 1% per degree Celsius.
#
# Task 3.2 – Thermal/slosh test:
# The test sets temperature to 28°C and tilt rate to 0.15 rad/s. Lower viscosity
# increases the predicted wave height, which should exceed the mug's slosh limit.
#
# Task 4.1 – Energy:
# The Warrant Gate rejects a proposal if estimated battery drain is not below
# 15% or if the remaining battery level is insufficient. The drain equation
# in this implementation is a simple modelling assumption, not a formula
# specified in the task text.
#
# Task 4.2 – Low battery:
# With battery level set to 5%, the robot should request recharging instead
# of attempting coffee delivery.
#
# Task 5.1 – Audit analysis:
# audit_analyzer counts approved and rejected proposals, counts failed warrant
# criteria, and compares predicted force with actual force for executed actions.
#
# Task 5.2 – Epistemic gap:
# 'asked' stores the original instruction and interpreted goal; 'belief' stores
# the robot's internal world estimate; 'observed' stores raw sensor evidence;
# 'committed' stores the proposed action and warrant decision; and
# 'execution_outcome' stores what happened physically. Keeping these separate
# makes it possible to compare intention, belief, sensor evidence and outcome.
```
